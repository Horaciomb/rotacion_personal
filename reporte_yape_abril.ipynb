```python
import pandas as pd

# Leer el Excel
df = pd.read_excel("PERSONAL_YAPE.xlsx")

# Normalizar columnas
df["FECHA DE INGRESO"] = pd.to_datetime(df["FECHA DE INGRESO"], errors="coerce")
df["FECHA DE SALIDA"] = pd.to_datetime(df["FECHA DE SALIDA"], errors="coerce")
df["ESTADO (ACTIVO/INACTIVO)"] = df["ESTADO (ACTIVO/INACTIVO)"].str.strip().str.upper()

# Rango abril 2026
inicio_abril = "2026-04-01"
fin_abril = "2026-04-30"

# Activos
df_activos = df[df["ESTADO (ACTIVO/INACTIVO)"] == "ACTIVO"]

# Ingresos abril
df_ingresos_abril = df[(df["FECHA DE INGRESO"] >= inicio_abril) & (df["FECHA DE INGRESO"] <= fin_abril)]

# Salidas abril
df_salidas_abril = df[(df["FECHA DE SALIDA"] >= inicio_abril) & (df["FECHA DE SALIDA"] <= fin_abril)]

# Distribución por ciudad
ciudad_counts = df_activos["CIUDAD"].value_counts()

# Resultados
print("="*60)
print(f"Cantidad de personas ACTIVAS: {len(df_activos)}")
print(f"Cantidad de personas que ingresaron en abril 2026: {len(df_ingresos_abril)}")
print(f"Cantidad de personas que salieron en abril 2026: {len(df_salidas_abril)}")
print("="*60)

print("\nDistribución por CIUDAD:")
print(ciudad_counts)
```

```python
import pandas as pd

# Leer el Excel
df = pd.read_excel("PERSONAL_YAPE.xlsx")

# Normalizar columnas
df["ESTADO (ACTIVO/INACTIVO)"] = df["ESTADO (ACTIVO/INACTIVO)"].str.strip().str.upper()
df["FECHA DE INGRESO"] = pd.to_datetime(df["FECHA DE INGRESO"], errors="coerce")
df["FECHA DE SALIDA"] = pd.to_datetime(df["FECHA DE SALIDA"], errors="coerce")

# Definir rango abril 2026
inicio_abril = "2026-04-01"
fin_abril = "2026-04-30"

# Filtrar personas que ingresaron en abril
df_ingresos_abril = df[(df["FECHA DE INGRESO"] >= inicio_abril) & (df["FECHA DE INGRESO"] <= fin_abril)]

# Filtrar personas que salieron en abril
df_salidas_abril = df[(df["FECHA DE SALIDA"] >= inicio_abril) & (df["FECHA DE SALIDA"] <= fin_abril)]

# Contar activos e inactivos SOLO de abril (ingresos + salidas)
df_abril = pd.concat([df_ingresos_abril, df_salidas_abril]).drop_duplicates()

activos_abril = (df_abril["ESTADO (ACTIVO/INACTIVO)"] == "ACTIVO").sum()
inactivos_abril = (df_abril["ESTADO (ACTIVO/INACTIVO)"] == "INACTIVO").sum()

total_abril = activos_abril + inactivos_abril
rotacion_abril = (inactivos_abril / total_abril) * 100 if total_abril > 0 else 0

print(f"Activos en abril 2026: {activos_abril}")
print(f"Inactivos en abril 2026: {inactivos_abril}")
print(f"Total en abril 2026: {total_abril}")
print(f"Porcentaje de rotación en abril 2026: {rotacion_abril:.2f}%")
```
